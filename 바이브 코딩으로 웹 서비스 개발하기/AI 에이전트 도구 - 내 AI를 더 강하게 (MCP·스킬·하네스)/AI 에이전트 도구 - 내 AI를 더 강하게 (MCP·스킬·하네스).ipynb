## 어시스턴트에서 에이전트로 - 도구의 정체

### AI가 내 서비스에서도 직접 일하려면

내 서비스에서 저장 오류가 나서 AI에게 "왜 저장이 안 되는지 찾아서 고쳐 줘"라고 묻는 장면을 떠올려 보세요. AI가 파일과 데이터베이스에 접근할 수 없다면, 내가 코드와 오류 메시지를 복사해 주고 수정 방법도 직접 적용해야 합니다. 답을 잘하는 AI가 있어도 내 서비스 안에서 직접 일하지 못하니 답답하죠. AI가 필요한 도구를 쓰고 결과까지 확인하게 하려면 무엇을 갖춰야 할까요?

### 오늘의 핵심 한 문장

> **AI 에이전트는 '모델 + 하네스'입니다. MCP나 앱의 CLI로 내 데이터·도구에 연결하고, 규칙 파일(AGENTS.md·CLAUDE.md)로 프로젝트 안내를 상시 걸어 두고, 스킬로 반복 작업을 외워 두면, 범용 AI가 '내 서비스 전용 조수'로 바뀝니다. 도구는 계속 바뀌지만, 이 개념은 남습니다.**

### 답하는 AI에서 행동하는 AI로

에이전트를 한 문장으로 줄이면, 목표를 주면 필요한 도구를 골라 쓰고 결과를 보며 다음 수를 정하기를 끝까지 반복하는 AI입니다. 우리가 하는 건 "이렇게 저렇게 클릭해"가 아니라 "로그인 기능을 붙여 줘"처럼 목표를 주는 것뿐이고, 세부 단계는 AI가 알아서 짭니다. 게다가 이 녀석은 말로만 답하지 않아요. 파일을 고치고, 명령을 실행하고, 검색을 하며 실제로 움직입니다. 그리고 한 번에 끝내는 대신 해보고, 결과를 확인하고, 틀리면 고치기를 목표가 이뤄질 때까지 되풀이하죠.

이건 코딩에만 해당하는 얘기가 아닙니다. 고객 문의를 읽고 답하는 AI든, 자료를 뒤져 보고서를 쓰는 AI든 뼈대는 똑같아요. 바이브 코딩 도구에서 파일을 고치고 실행 결과를 확인하는 AI가 바로 코딩 에이전트입니다.

'어시스턴트'와 헷갈릴 때가 있는데, 둘은 칼로 자르듯 나뉘기보다 정도의 차이입니다.

| 구분 | 어시스턴트(assistant) | 에이전트(agent) |
|---|---|---|
| 하는 일 | 다음 줄 추천, 질문에 답 | 여러 파일 수정·명령 실행·반복 |
| 단계를 정하는 주체 | 사람 | AI가 스스로 |
| 도구 사용 | 거의 없음(텍스트 위주) | 파일·터미널·검색 등 |
| 비유 | 곁에서 거드는 조수 | 일을 맡아 끝내는 일꾼 |

표의 오른쪽, 그러니까 스스로 단계를 정하고 도구까지 써서 일을 끝내는 쪽이 우리가 말하는 에이전트입니다. (출처: Anthropic, "Building effective agents")

### 잠깐, 왜 '에이전트'라고 부를까

사실 이 단어는 우리한테 낯설지 않습니다. 부동산 에이전트, 여행사 에이전트, 첩보 영화 속 요원(agent)까지, 다 '나 대신 뛰어서 일을 처리해 주는 사람'을 가리키죠. 어원도 그렇습니다. 라틴어 '행하다(agere)'에서 나와 수백 년 전부터 "남을 대신해 행동하는 이"라는 뜻으로 쓰였어요. AI가 지어낸 신조어가 아닙니다.

인공지능 쪽에서도 이 말은 한참 전부터 단골이었습니다. AI 교과서의 고전인 러셀·노빅의 책(1995)은 인공지능이라는 학문을 아예 "주변을 살펴보고 목표를 위해 스스로 행동하는 존재, 즉 에이전트를 연구하는 일"이라고 정의해 버렸을 정도예요. 1990년대에 이미 내 메일을 읽고 일정을 잡아 주던 비서 프로그램을 '에이전트'라 불렀고요. '스스로 감지하고 행동한다'는 알맹이는 그때 다 있었던 겁니다.

그런데 왜 하필 요즘 이 옛말이 유행어가 됐을까요. 얼마 전까지 에이전트는 주로 연구실 용어였습니다. 그러다 2023년 봄, "목표만 던지면 알아서 끝까지 해보겠다"는 LLM 실험들이 화제가 되면서 일반 사람들 입에도 오르내리기 시작했죠(이 소동은 잠시 뒤에 다시 봅니다). 뜻의 뿌리는 예나 지금이나 같습니다. 달라진 건 그 '머리' 역할을 이제 LLM이 맡는다는 것뿐이고요.

*(출처: 스튜어트 러셀·피터 노빅, 『인공지능: 현대적 접근』, 1995; 파티 마스, 1994)*

### 핵심 공식: 에이전트 = 모델 + 하네스

2026년 도구들을 이해하는 열쇠가 되는 공식이 있습니다.

> **에이전트 = 모델(model) + 하네스(harness)**

- **모델** 은 언어·코드·이미지를 다루는 **지능** 입니다 (Claude, GPT 같은 것).
- **하네스** 는 그 지능을 감싸서 **실제 일을 하게 만드는 실행 환경** 입니다 - "읽고 → 고치고 → 실행하고 → 다시 반복"하는 루프, 파일 접근, 안전장치 등이 여기에 들어갑니다.

모델의 능력과 가격·속도는 계속 달라지고, 같은 모델도 하네스의 도구 권한·검증·사용 흐름에 따라 결과가 크게 달라집니다. 그러니 "모델이 다 비슷하다"고 단정하기보다, **내 작업에서 어떤 하네스가 모델을 안전하고 확인 가능하게 움직이는가** 를 비교해 봅시다. (출처: DEV, "Every AI Coding CLI in 2026")

오늘은 이 하네스를 **더 강하게 만드는 세 가지 방법(MCP·스킬·규칙)** 과, **어떤 하네스를 고를지** 를 살펴봅니다.

### 역사 이야기: AI가 '답하기'에서 '스스로 일하기'까지

에이전트는 어느 날 뚝 떨어진 게 아니라, AI와 우리가 대화하는 방식이 몇 년에 걸쳐 조금씩 바뀌며 도착한 자리입니다.

처음(2020～2021년)의 AI는 한 번 묻고 한 번 답하면 끝이었습니다(이른바 '단일 턴'). 초기 GPT 같은 모델은 문장을 던지면 그 뒤에 이어질 말을 한 덩이 지어 줄 뿐, 방금 나눈 얘기도 기억하지 못했죠. 그 무렵 코드 편집기에서는 다음 줄을 대신 채워 주는 자동완성(GitHub Copilot) 정도가 전부였습니다.

2022년 말 ChatGPT가 나오며 분위기가 바뀝니다. 앞서 주고받은 대화를 기억하고 이어 가는 '멀티턴' 대화가 되자 "코드를 대화로 부탁한다"는 방식이 열렸어요. 그래도 이때까지 AI는 말로만 거들었습니다. 스스로 뭘 실행하진 못했죠.

진짜 전환은 AI가 말 대신 행동을 하게 된 순간입니다. 2023년 들어 모델이 정해진 방식으로 바깥 도구를 부를 수 있게 되면서, 검색을 하고 파일을 읽고 코드를 실행하기 시작했습니다. 이때부터 AI는 '낯선 천재'에서 '내 코드베이스를 좀 아는 동료'로 넘어왔습니다.

여기에 '반복'이 얹히자 에이전트가 됩니다. 2024년 이후로는 목표만 주면 계획하고, 실행하고, 결과를 보고 고치기를 스스로 되풀이하는 도구(Cursor나 Claude Code 같은)가 자리를 잡았고, 이슈를 통째로 맡기면 알아서 코드와 테스트까지 가져오는 데까지 왔습니다.

한마디로 AI는 답만 하던 데서 대화를 하고, 도구를 쓰고, 이제 스스로 반복하는 데까지 온 겁니다. 겉모습은 도구마다 달라도 속은 점점 닮아 가는데, 그래서 `AGENTS.md`나 스킬, MCP처럼 '한 번 적어 두고 계속 쓰는 맥락'이 어느 도구에서나 공통으로 등장하죠. (출처: CodeRabbit, "A brief history of AI coding")

### 작동 원리: 에이전트는 어떻게 '스스로' 일하나 - 루프

챗봇과 에이전트의 결정적 차이는 **반복(루프)** 입니다. 챗봇은 한 번 묻고 한 번 답하지만, 에이전트는 목표가 끝날 때까지 같은 사이클을 돕니다.

- **계획(Plan)** - 뭘 할지 정하고, 큰 목표를 작은 일로 쪼갠다
- **실행(Act)** - 실제로 한다 (코드 실행·파일 수정·명령 실행)
- **관찰(Observe)** - 결과를 본다 (에러가 났나? 됐나?)
- **반성(Reflect)** - 제대로 됐는지 판단하고, 아니면 다시 계획으로

이 네 단계를 **목표를 이룰 때까지** 반복합니다. 프로그래밍의 '반복문(while)'처럼요. 이 '생각하고-행동하고-관찰한다'는 발상을 연구자들은 **ReAct** 라 부릅니다. 이 방식이 한 방에 답하는 것보다 실제로 성능이 좋다는 게 실험으로 확인됐습니다.

> 사실 새로운 발상은 아닙니다 - 품질관리의 **PDCA(계획-실행-점검-개선)**, 공군 전략의 **OODA 루프** 가 다 같은 모양입니다. "계획하고, 해 보고, 결과를 보고, 고친다." 바이브 코딩에서도 앱의 오류를 확인하고, 코드를 수정하고, 다시 실행해 결과를 살피는 과정을 반복합니다. 에이전트는 이 오래된 지혜를 **기계 속도로** 돌리는 것입니다. (출처: ikangai, "The Agentic Loop Explained"; Yao et al., "ReAct", 2022)

### 작동 원리: '생각하고 → 행동하고 → 다시 생각한다'의 뿌리 - ReAct

에이전트가 스스로 도는 그 루프의 대표적 연구가 **ReAct** 입니다. Shunyu Yao와 공동저자들은 2022년 10월 arXiv에 논문을 공개했고, 2023년 ICLR에 발표했습니다. 저자 소속에는 Princeton University와 Google Research가 포함됩니다.

이름은 **Reasoning(추론) + Acting(행동)** 을 합친 것입니다. 모델이 혼자 답을 완성하는 대신, **"생각을 적고 → 실제로 행동(예: 검색)하고 → 관찰한 결과를 보고 → 다시 생각하기"** 를 번갈아 합니다. 사람이 "이 사실은 확인해야겠다 → 찾아봄 → 그 결과로 다음 선택을 정함"이라고 푸는 과정과 닮았습니다.

> 오늘날 모든 코딩 에이전트가 ReAct를 그대로 구현한다는 뜻은 아닙니다. 다만 파일을 열고, 명령을 실행하고, 결과를 근거로 다음 행동을 정하는 **관찰-수정 루프** 를 이해하는 좋은 출발점입니다. (출처: Yao et al., "ReAct", arXiv:2210.03629, 2022; ICLR 2023)

![ReAct 루프 인포그래픽: Thought, Action, Observation이 순환한다](https://ebhtmardjzsiyxjncvbb.supabase.co/storage/v1/object/public/images/01-vibe-coding-web/day09_react_loop.png)
*생각만으로 끝내지 않고, 행동의 결과를 다음 판단의 근거로 삼는 ReAct 루프.*

### ReAct 루프: 왜 '한 번에 정답'보다 나을까

**Thought** 는 다음에 확인할 가설과 계획을 세우는 단계입니다. 그러나 계획은 파일의 실제 내용이나 명령의 오류를 모른 채 세운 추측일 수 있습니다. **Action** 은 파일 읽기·검색·테스트 실행처럼 바깥 세계에 질문을 던지는 단계이고, **Observation** 은 그 질문이 돌려준 사실입니다. 다음 Thought는 그 사실을 재료로 다시 쓰입니다.

예를 들어 "로그인이 안 된다"면 에이전트가 곧바로 코드를 고치지 않고, 먼저 오류 로그를 읽고(Action), `401`(인증 실패)인지 환경 변수(프로그램 설정값) 누락인지 관찰합니다(Observation). 그 뒤에만 수정 방법을 고릅니다(Thought). 이 교차가 환각을 없애지는 않지만, **추측을 검증 가능한 증거로 교체** 하게 합니다.

따라서 좋은 지시는 "고쳐 줘"에서 끝나지 않습니다. "먼저 관련 파일과 오류를 확인하고, 수정 뒤에는 테스트를 실행해 결과를 보고해 줘"처럼 관찰 지점을 요구합니다. 루프가 길어질수록 비용과 실수 가능성도 늘므로, 사람이 멈출 기준을 함께 정해야 합니다.

### 역사 이야기: 2023년 봄의 소동 - "목표만 주면 알아서" AutoGPT

에이전트라는 말이 대중에게 확 퍼진 건 **2023년 봄** 입니다. GPT-4가 나온 직후, **"목표 한 줄만 주면 알아서 할 일을 쪼개 끝까지 해내는"** 프로그램들이 화제가 됐습니다. **BabyAGI**(투자자 요헤이 나카지마가 만든 짧은 스크립트)와 **AutoGPT** 가 대표였습니다. AutoGPT 저장소는 출시 몇 주 만인 **2023년 4월 말～5월 초에 GitHub 별 10만 개를 넘겨** 당시 사상 최고 속도로 불렸습니다. 이는 폭발적 관심을 보여 주는 지표이지, 품질이나 현재 사용성을 보장하는 순위는 아닙니다.

그런데 실제로 돌려 보니 문제가 드러났습니다 - **언제 멈춰야 할지 몰라 같은 일을 무한 반복** 하고, 엉뚱한 방향으로도 계속 AI를 호출해 **요금이 커지며**, 중간 결과를 사람이 확인하지 않으면 틀린 목표를 향해 달릴 수 있었습니다.

그래서 오늘의 에이전트는 "완전 자율"보다 **작은 작업 단위·명확한 권한·중간 확인** 을 더 중요하게 봅니다. 멋진 자율성보다, 내가 멈추고 검토할 수 있는 작업 흐름이 실전에서는 더 안전합니다. (출처: AutoGPT 저장소 및 당시 보도, 아래 출처 목록)

### 내 데이터에 접근할 통로가 필요합니다

에이전트가 결과를 확인하며 일하려면, 먼저 필요한 파일이나 서비스에 접근할 수 있어야 합니다. 예를 들어 저장 오류를 조사할 때는 오류 메시지뿐 아니라 실제 데이터베이스의 상태도 확인해야 하죠. AI 앱과 외부 데이터·도구가 공통된 방식으로 연결되도록 돕는 규격이 MCP입니다.

### 문제: AI는 '내 것'을 모릅니다

아무리 똑똑한 AI도 기본적으로는 **자기가 학습한 것만** 압니다. 내 데이터베이스, 내 파일, 내 캘린더, 우리 회사 문서는 모릅니다. 그때그때 복사해서 붙여넣을 수도 있지만, 연결할 도구가 많아지면 **"AI 3개 × 도구 5개 = 연결 15개"** 를 일일이 만들어야 하는 지옥이 됩니다. (이걸 M×N 문제라고 부릅니다.)

### MCP = "AI를 위한 USB-C 포트"

**MCP(Model Context Protocol)** 는 이 문제를 푸는 **표준 규격** 입니다. Anthropic이 2024년 말에 공개했고, 지금은 업계 표준이 되었습니다.

> USB-C를 떠올려 보세요. 기기마다 다른 단자를 쓰는 대신 공통 단자를 쓰면 연결이 쉬워집니다. 다만 모양이 같아도 필요한 기능과 전력 규격이 맞아야 하죠. **MCP는 'AI를 위한 USB-C'** 입니다. 도구를 MCP 규격으로 만들어 두면 여러 AI 앱에서 재사용할 수 있지만, **서로 지원하는 전송 방식·프로토콜 버전·인증 방식이 맞아야** 연결할 수 있습니다. (출처: modelcontextprotocol.io, "What is MCP"; IBM, "What is Model Context Protocol?")

무엇을 할 수 있냐면 -

- AI가 **내 데이터베이스를 직접 조회** 해서 답하기
- **Figma 디자인** 을 읽어서 그대로 화면 만들기
- 내 **캘린더·Notion·Slack** 에 접근해 개인 비서처럼 굴기

### 역사 이야기: MCP는 어쩌다 '표준'이 됐나

MCP가 정말 흥미로운 건 **경쟁사들이 같은 규격에 손을 얹었다** 는 점입니다. 앞서 본 'M×N 통합 지옥'을 풀 공용 규격이 필요했고, Anthropic이 2024년 11월 공개했는데, 이듬해 **OpenAI(2025년 3월)와 구글(2025년 4월)이 잇따라 채택** 했습니다. 그리고 2025년 말에는 여러 회사가 함께 관리하는 중립 재단(리눅스 재단 산하 Agentic AI Foundation)으로 넘어갔습니다. 경쟁하는 회사들이 같은 규격을 함께 떠받치는 건 흔치 않은 일입니다. (출처: Anthropic, "Model Context Protocol", 2024; TechCrunch, 2025)

> 그래서 오늘 배우는 MCP는 특정 회사 기능이 아니라 **업계 공용 약속** 입니다 - 한 번 익히면 도구가 바뀌어도 계속 쓸 수 있다는 뜻이죠.

### MCP로 실제로 뭘 할까 - 사례 모음

MCP가 추상적으로 느껴질 수 있으니, 사람들이 실제로 연결해 쓰는 예를 봅시다.

- **디자인 → 코드**: Figma를 MCP로 연결하면, AI가 디자인을 **읽어서 그대로 화면** 을 만듭니다.
- **내 문서와 대화**: Notion을 연결해 "이번 주 할 일 정리해 줘", "이 문서 요약해 줘"를 채팅 안에서.
- **웹을 데이터로**: 웹사이트를 훑어 **상품·가격·이미지** 를 정리해 오기(스크래핑)를 자연어로.
- **글을 소리로**: 긴 글을 음성으로 바꾸기.
- **일정 자동화**: "다음 주 화요일 회의 잡아 줘" → 캘린더에 초대장.
- **사내 업무**: 회사 DB·내부 문서·ERP에 붙여 IT 지원·채용·계약 업무를 돕기.

Anthropic이 Google Drive·Slack·GitHub·Postgres 같은 **기성 MCP 서버** 를 공개해 뒀고, 개발 도구들(Cursor·Zed·Replit 등)도 MCP를 받아들였습니다.

> 흥미로운 현실 - 공개적으로 널리 쓰이는 MCP 서버는 아직 손에 꼽지만, **회사 내부용** MCP(사내 DB·문서 연결)가 가장 값지다는 평가입니다. 남에게 공개하지 않아도, **내 데이터에 AI를 붙이는 것** 만으로 큰 힘이 되니까요. (출처: Merge, "MCP integration examples"; Pragmatic Engineer, "Building MCP servers")

### 실전 MCP - 사람들이 실제로 많이 붙이는 것들

MCP 서버는 이제 '앱스토어'처럼 한곳(공식 레지스트리)에서 찾습니다. 여러분이 "내 작업엔 뭘 붙일까" 감을 잡도록 대표적인 것만 추렸습니다.

| MCP 서버 | 한 줄 용도 | 이런 사람에게 |
|---|---|---|
| **파일시스템** | 내 폴더의 파일을 읽고 쓰기 | 문서·자료 정리하는 누구나 |
| **GitHub** | 저장소 이슈·코드 읽고 커밋·검색 | 프로젝트를 GitHub에 올리는 사람 |
| **브라우저 자동화(Playwright)** | AI가 실제 웹브라우저를 열어 클릭·입력·캡처 | 반복 웹작업·테스트 |
| **Slack** | 채널 내역 검색·메시지 보내기 | 팀 소통 자동화 |
| **Notion** | 노션 페이지·DB 읽고 쓰기 | 노션으로 지식 관리하는 사람 |
| **데이터베이스** | DB에 자연어로 물어 조회(읽기 전용 권장) | 앞서 배운 내 서비스 DB와 연결 |
| **구글 드라이브·캘린더** | 문서 검색·일정 확인/생성 | 문서·일정 자동화 |

(출처: 공식 MCP 레지스트리 https://registry.modelcontextprotocol.io)

> 딱 하나 조심 - 공개된 MCP 서버는 수천 개 이상으로 급증했고 그중엔 검증 안 된 것도 많습니다. **모르는 서버를 함부로 붙이지 말고, 회사가 직접 낸 공식 서버 위주로** 시작하세요. AI에게 내 데이터·계정 접근 권한을 주는 일이니까요.

### 개념 정리: MCP가 건네는 세 가지 - 도구·자원·프롬프트

MCP로 연결하면 AI가 세 종류를 쓸 수 있게 됩니다. 이름만 알아 두면 됩니다.

- **도구(Tools)** - AI가 **실행** 하는 기능 (예: 검색하기, DB 조회하기)
- **자원(Resources)** - AI가 **읽어 들이는** 데이터 (예: 파일, 문서, 기록)
- **프롬프트(Prompts)** - 미리 만들어 둔 **요청 템플릿**

이게 왜 좋을까요? ① 연결을 매번 새로 안 만들어도 되고, ② AI가 **실제 최신 데이터** 를 보니 **엉뚱한 답(환각)이 줄고**, ③ 학습 시점 이후의 최신 정보도 쓸 수 있습니다. (출처: modelcontextprotocol.io; Descope, "What Is MCP")

> MCP는 결국 이렇게 다가옵니다 - "AI야, **우리 서비스 DB** 보고 답해 줘"가 가능해지는 것. 직접 연동 코드를 짜지 않더라도, 사용할 MCP 연결을 AI 앱 안내에 따라 설치하거나 등록한 뒤 필요한 인증·로그인을 마치고 요청된 권한을 확인해 승인하면, 앱에 도구 목록이 나타나고 읽기 요청이 정상 작동하는지 확인해 연결을 시작할 수 있습니다.

*Host의 클라이언트와 서버가 표준 방식으로 대화하고, 서버가 세 종류의 기능을 제공하는 구조.*

### MCP의 약속: 연결 하나를 다시 만들지 않기

MCP에서 **host** 는 AI를 담은 앱(예: 코딩 도구)이고, 그 안의 **client** 가 연결을 관리합니다. **server** 는 Notion·파일 시스템·DB 같은 바깥 기능을 MCP 방식으로 내놓습니다. server가 제공하는 것은 실행 가능한 **tools**, 읽을 수 있는 **resources**, 재사용할 **prompts** 입니다. 한 서버가 무엇을 허용하는지는 서버가 명시하고, host는 그 목록을 AI에게 보여 줍니다.

표준이 없으면 AI 앱 M개와 데이터 서비스 N개가 각각 전용 연동을 만들어야 하므로 연결이 **M×N개** 가 됩니다. 공통 규약을 따르면 앱은 MCP client를 한 번 구현하고, 서비스는 MCP server를 한 번 구현해 조합할 수 있어 대략 **M+N개** 의 구현으로 줄어듭니다. USB-C 비유가 여기서 나옵니다. 플러그 모양이 같다고 모든 기기가 같은 권한을 갖는 것은 아닌 것처럼, MCP도 연결 규칙이지 자동 안전 보장은 아닙니다.

특히 tool은 실제 행동을 할 수 있습니다. 처음에는 "읽기" 도구부터 연결해 접근을 허용한 데이터가 조회되는지 확인합니다. 로그인에 성공했다고 모든 데이터에 접근할 수 있는 것은 아니므로, 승인한 권한 범위도 함께 살펴봅니다. 쓰기·삭제·배포처럼 되돌리기 어려운 권한은 승인 과정을 둡니다. AI가 최신 데이터를 읽는다고 해서 해석까지 자동으로 맞는 것은 아니므로, 중요한 결과는 원본과 함께 확인합니다.

## 스킬 - 반복 작업을 AI에게 외워 두기

## 스킬(Skill) - 반복하는 일을 AI에게 '외워' 두기

---

### 문제: 매번 같은 설명을 다시 하고 있지 않나요?

AI에게 "우리 브랜드 톤으로 정리해 줘", "이 형식으로 보고서 만들어 줘"를 **매번 처음부터 설명** 하고 있다면 - 지겹고, 그때그때 결과도 달라집니다. 이걸 **재프롬프트(re-prompting) 문제** 라고 합니다.

### 스킬 = 재사용하는 '플레이북'

**스킬(Skill)** 은 자주 하는 작업을 **한 번 적어 두고 계속 꺼내 쓰는 설명서** 입니다. 기술적으로는 그냥 **폴더 하나** 입니다 - 그 안에 `SKILL.md`라는 파일에 **"무엇을, 언제 쓰는지"** 를 적어 두면, AI가 관련된 일을 만났을 때 **알아서 꺼내 씁니다.** (출처: Claude Platform Docs, "Agent Skills")

핵심은 **필요한 설명만 그때그때 읽는 구조** 입니다. AI는 평소엔 스킬의 **이름과 짧은 설명만** 읽고, 작업에 필요할 때 본문을 읽습니다. 이름과 설명이 약 30∼50 토큰인 경우를 생각해 볼 수 있지만, 실제 분량은 작성 방식에 따라 달라집니다. '토큰'은 AI가 글을 읽고 쓰는 단위이며 요금 계산에도 쓰입니다. 자세한 건 뒤 하네스 섹션에서 봅니다. 본문을 항상 읽는 방식보다 입력량을 줄일 수 있지만, 스킬이 많아지면 이름과 설명을 읽는 양도 늘어나므로 **비용이나 응답 시간이 늘지 않는다는 뜻은 아닙니다.**

> 예를 들면 - "내가 쓴 글을 **우리 서비스 말투로** 다듬는 법"을 스킬로 만들어 두면, 다음부터는 "말투 스킬 써서 다듬어 줘" 한마디면 됩니다.

### SKILL.md는 실제로 어떻게 생겼나

스킬 폴더 안 `SKILL.md`는 두 부분입니다 - 맨 위 **머리말(이름·설명)** 과 그 아래 **본문(실제 지시)**. 예를 들어 "우리 서비스 말투로 안내문 다듬기" 스킬은 이렇게 생겼습니다.

```markdown
---
name: tone-polish
description: 사용자용 안내 문구를 우리 서비스 말투(정중하고 짧게)로 다듬을 때 사용
---

# 우리 서비스 말투로 다듬기

1. 한 문장은 40자 이내로 끊는다.
2. "～해야 합니다"보다 "～해 주세요"처럼 부탁형으로 쓴다.
3. 전문용어는 괄호로 짧게 풀어 준다.
4. 마지막에 도움받을 곳(고객센터 링크)을 붙인다.
```

핵심 원리는 **점진 공개(progressive disclosure)** 입니다. AI는 평소 머리말의 `name`과 `description`만(수십 토큰 수준) 훑고 있다가, 지금 하는 일이 그 설명과 맞을 때만 본문 전체를 읽어 옵니다. 본문을 필요할 때만 읽으므로 처음부터 모든 설명을 넣는 것보다 입력량을 줄일 수 있습니다. 다만 머리말을 읽는 데도 토큰이 들며, 실제 비용과 응답 시간은 스킬의 수·길이와 도구의 동작에 따라 달라집니다. `description`을 잘 쓰는 게 중요한 이유가 여기 있습니다 - AI가 **언제 이 스킬을 꺼낼지** 를 그 한 줄로 판단하니까요. (출처: Claude Platform Docs, "Agent Skills")

### 남들이 이미 만들어 둔 스킬들

스킬은 내가 직접 쓸 수도 있지만, 잘 만들어진 걸 가져다 쓰는 게 더 빠를 때가 많습니다. 가장 유명한 건 Anthropic이 만든 '문서 스킬' 네 개예요. 파워포인트(pptx), 엑셀(xlsx), 워드(docx), PDF. claude.ai에서 "이 자료로 발표 슬라이드 만들어 줘"라고 하면 뒤에서 조용히 이 스킬들이 돌아가는데, 마크다운을 흉내 낸 가짜가 아니라 서식까지 갖춘 진짜 .pptx·.xlsx 파일을 뽑아 줍니다. 사실 Claude의 문서 생성 기능이 통째로 이 스킬들 위에서 돌아가고, 소스도 공개돼 있어서 '잘 만든 스킬은 어떻게 생겼나'의 교본으로 자주 인용됩니다.

그 밖에도 쓸 만한 게 많습니다. 새 스킬을 만들 때 형식을 잡아 주는 '스킬 만드는 스킬', 앞서 배운 MCP 서버를 설명만 주면 짜 주는 'MCP 빌더', 브랜드 가이드라인을 지키게 하는 스킬 같은 것들이죠. Anthropic [공식 저장소](https://github.com/anthropics/skills)에 예시가 모여 있고, 커뮤니티가 올린 것도 수백 개입니다.

딱 하나 조심할 것. 스킬은 결국 'AI에게 주는 지시문 + 실행 코드'라, 출처가 불분명한 걸 함부로 켜면 위험합니다. 내가 만들었거나 공식 저장소에서 온 것 위주로 쓰고, 남의 스킬은 안에 뭐가 들었는지 한 번 열어 보고 쓰세요. (출처: Anthropic, "Agent Skills" 공식 문서; anthropics/skills 저장소)

### 매번 설명하기 싫다면 - 규칙 파일(AGENTS.md·CLAUDE.md)

스킬과 비슷하지만 더 기본이 되는 게 있습니다. AI에게 **"우리 프로젝트는 이런 규칙이야"를 매번 설명하기 지겨울 때**, 그걸 파일 하나에 적어 두는 것입니다. 흔히 **`AGENTS.md`** 또는 `CLAUDE.md`라고 부르고, 공식 설명은 이걸 **"에이전트를 위한 README"** 라고 합니다. 많은 도구가 대화를 시작할 때 이런 규칙 파일을 자동으로 읽어 프로젝트의 기본 지침으로 삼습니다. **Claude Code는 버전에 따라 `CLAUDE.md`와 `AGENTS.md`를 자동으로 읽습니다.** 지원 범위와 자동 로딩 조건이 다를 수 있으므로, 사용하는 버전에서 어떤 파일을 읽는지 확인하세요. (출처: agents.md 공식 스펙)

무엇을 적으면 유용할까요?

- **프로젝트 개요** - 이게 무슨 서비스인지 한두 줄
- **자주 쓰는 명령** - 실행·테스트 방법(설명보다 명령어를 먼저)
- **규칙·컨벤션** - 폴더 구조, 스타일 등
- **하지 말 것** - "이 폴더는 건드리지 마" 같은 금지사항

> 흐름 하나 - 예전엔 도구마다 파일 이름이 제각각이었는데, 2025년 여러 회사가 모여 **`AGENTS.md`라는 공통 표준** 을 만들었습니다(수만 개 프로젝트가 채택). 스킬이 '반복 작업의 플레이북'이라면, 컨텍스트 파일은 **'프로젝트의 상시 안내문'** 입니다. 짧고 명령어 중심으로 쓰는 게 좋습니다.

### 헷갈리지 않기 - MCP vs 스킬 vs 규칙

셋이 비슷해 보이지만 역할이 다릅니다. 이 결정 규칙이면 충분합니다. (출처: Totalum, "Claude Code Skills in 2026")

| 내가 원하는 것 | 무엇을 쓰나 |
|---|---|
| AI에게 **새 능력·연결** 을 주고 싶다 (내 DB·도구에 붙이기) | **MCP** |
| AI가 **항상 이런 식으로** 행동하게 하고 싶다 | **규칙** (프로젝트 설정 파일: CLAUDE.md / AGENTS.md) |
| **반복하는 다단계 작업** 을 자동화하고 싶다 | **스킬(Skill)** |

한 문장으로: **스킬은 선반에서 꺼내 쓰는 '방법서', 규칙은 프로젝트의 '상시 안내문', MCP는 도구를 꽂는 '어댑터'입니다.** 서로 경쟁이 아니라 함께 씁니다.

### 미니 퀴즈 - MCP? 스킬? 규칙?

아래 상황에 무엇이 맞을지 골라 보세요.

| 상황 | MCP / 스킬 / 규칙 |
|---|---|
| AI가 우리 회사 DB를 직접 조회하게 하고 싶다 |  |
| "보고서는 항상 우리 양식으로"를 늘 지키게 하고 싶다 |  |
| "경쟁사 조사 → 표로 정리"라는 다단계 작업을 매번 반복한다 |  |

<details>

- DB에 연결(새 능력) → **MCP**
- 항상 이렇게(행동 규칙) → **규칙**(CLAUDE.md / AGENTS.md)
- 반복하는 다단계 작업 → **스킬(Skill)**

새 능력=MCP, 항상=규칙, 반복 작업=스킬. 이 세 갈래만 기억하면 됩니다.

## 하네스 종류 비교 - 어떤 AI 코딩 도구를 고를까

---

### 도구를 고르기 전, 내 자리부터 떠올려 보기

마감 전날, 여러분은 브라우저에서 빠르게 화면을 손보고 싶을 수 있습니다. 반대로 팀의 저장소를 내려받아 테스트와 터미널 명령까지 맡기고 싶을 수도 있습니다. 둘 다 "AI 코딩"이지만 필요한 하네스와 권한은 다릅니다. 표를 보기 전에 **내가 어디서 일하고, 무엇을 AI에게 맡길지** 를 먼저 정해 봅시다.

비교할 때는 모델 이름만 보지 말고 네 가지를 확인합니다. 파일을 읽고 쓸 수 있는지, 터미널·브라우저 같은 도구를 쓸 수 있는지, MCP로 내 자료를 연결할 수 있는지, 실행 전후에 내가 승인·검토할 수 있는지입니다. 처음에는 되돌리기 쉬운 작은 수정으로 시험해 보는 것이 가장 좋은 비교 방법입니다.

"에이전트 = 모델 + **하네스**"라고 했죠. 이 하네스(도구)는 종류가 아주 많지만, 사용 방식으로 보면 **네 부류** 로 정리할 수 있습니다. 다만 화면의 형태와 일을 맡기는 방식이 함께 들어 있는 분류라, 한 도구가 여러 부류에 걸치기도 합니다. (출처: Faros AI, 2026; DEV, "Every AI Coding CLI in 2026")

- **터미널/CLI형** - 검은 명령창에서 돌아갑니다. 깊고 강력하지만 조금 낯섦. 예: **Claude Code**, Codex CLI, **Antigravity CLI**(agy), Aider, OpenCode
- **편집기(IDE)형** - VS Code 같은 코드 편집기 안에서. 예: **Cursor**, **Devin Desktop**(Windsurf에서 개명), Cline. (Roo Code 확장은 2026-05-15에 종료됨.)
- **브라우저형** - 웹에서 클릭 몇 번으로, **처음엔 가장 수월함**(배포까지 한 번에). 예: **Lovable**, Bolt, Replit Agent
- **클라우드/비동기형** - "이 이슈 고쳐 줘"를 통째로 맡기면 백그라운드에서 일하고 결과를 초안으로 가져옵니다. 예: **Devin**, Google Jules, OpenAI Codex(클라우드)

**어디서 실행되는지** 와 **내가 얼마나 지켜보는지** 는 따로 보아야 합니다. Lovable·Replit Agent는 내 브라우저로 조작하지만 작업은 클라우드에서 실행됩니다. 클라우드에서도 진행 상황을 보며 지시할 수 있고, 로컬 도구에도 긴 작업을 맡길 수 있습니다. 도구를 고를 때는 실행 장소와 함께, 중간에 확인하며 진행할지 맡긴 뒤 결과를 검토할지를 살펴보세요. 참고로 **Devin**(클라우드 자율 에이전트)과 **Devin Desktop**(구 Windsurf, 편집 환경)은 같은 회사(Cognition)의 제품군입니다. **Google Antigravity** 는 Google이 Gemini CLI를 이어받아 통합한 에이전트 도구입니다(2026-06 전환, 터미널은 `agy` 명령, 데스크톱 앱도 있음).

### 비교표 - 뭘 고를까 (2026년 · 커뮤니티 감안)

커뮤니티에서 가장 자주 나오는 결론부터 말하면 "정답 하나는 없다"입니다. 도구마다 성격이 뚜렷해서 고르는 사람의 자리와 취향을 타거든요. "코파일럿은 회사가 골라 주고, Claude Code나 Cursor는 엔지니어가 직접 고른다"는 우스개가 돌 정도죠. 대표적인 것들을 형태별로 늘어놓으면 이렇습니다.

| 도구 | 형태 | 강점 | 언제 / 누구에게 |
|---|---|---|---|
| **Claude Code** | 터미널 에이전트 | 깊은 추론·큰 코드베이스·긴 작업, 확장(MCP·스킬·훅)·에이전트 팀 | 어려운 문제를 끝까지 파고들 때 |
| **OpenAI Codex** | 터미널 + 클라우드 | 이슈를 백그라운드로 맡기고 결과를 받아 옴 | 돌려 두고 결과만 검토할 때 |
| **Google Antigravity**(agy CLI + 데스크톱) | 터미널·에이전트 환경 | Gemini CLI를 이어받은 Google 통합 도구(2026-06 전환) | Google 생태계로 작업할 때 |
| **Aider · opencode** | 터미널·오픈소스(BYOM) | 무료, 모델 골라 씀, git 자동 커밋 | 비용을 직접 조절하고 터미널이 편할 때 |
| **goose · Pi** | 터미널·오픈소스 에이전트 | 무료·모델 골라 씀, goose는 MCP·확장 풍부(리눅스재단 AAIF), Pi는 미니멀(OpenClaw의 엔진) | 오픈소스 터미널 에이전트를 직접 손보고 싶을 때 |
| **Cursor** | 편집기(VS Code 포크) | 편집 경험이 가장 매끄럽고 자동완성·비주얼 diff가 강함 | 편집기에서 매끄럽게 쓰고 싶을 때 |
| **Devin Desktop**(구 Windsurf) | 편집기/데스크톱 | 로컬 에이전트와 편집 환경을 함께 | 편집기 안에서 에이전트를 쓰고 싶을 때 |
| **Zed** | 편집기(고속·무료 티어) | 빠르고 가볍고 모델을 가리지 않음 | 가벼운 편집기를 원할 때 |
| **GitHub Copilot** | 여러 편집기의 확장 | 가장 널리 쓰이고 기업 보안·감사·정책이 성숙, 저렴 | 회사 규정·컴플라이언스가 중요할 때 |
| **Cline · Kilo Code** | VS Code 확장·오픈소스(BYOK) | 무료, 모델 골라 씀, 계획/실행 분리(Plan·Act) | 비용·모델·프라이버시를 직접 쥐고 싶을 때 |
| **OpenClaw · Hermes** | 자기개선·개인 비서형(오픈소스) | 챗앱으로 부리고, 기억을 남기며 스스로 스킬을 만듦 | 늘 켜 두는 '내 에이전트'를 원할 때 |
| **Lovable · Bolt · Replit** | 브라우저(노코드) | 클릭으로 만들고 바로 배포 | 코딩이 처음이거나, 기획자의 첫 도구 |
| **Devin · Google Jules** | 클라우드/비동기 | 이슈를 통째로 맡기면 알아서 처리 | 결과만 검토하며 맡기고 싶을 때 |

> 고르는 기준은 프레임워크 고를 때와 같습니다 - 정답이 아니라 내 조건. 터미널이 편한가 편집기가 편한가, 비용은, 확장(MCP·스킬)이 필요한가, 빠른 시제품인가 어려운 문제인가. "다들 최고라는 것" 말고 **내 상황** 으로 고르세요. (출처: Faros AI, "Best AI Coding Agents 2026"; DEV, "Every AI Coding CLI in 2026"; 커뮤니티 종합)

### 코드를 직접 다루는 도구 - 편집기·터미널·클라우드

앞에서 하네스를 **CLI·편집기·브라우저·클라우드** 네 부류로 봤죠. 그중 브라우저(노코드) 도구(Lovable·Bolt·Replit)는 **가장 쉬운 입문 갈래** 로 이미 살펴봤습니다. 여기서는 **코드를 직접 다루는** 나머지 세 갈래가 실제로 어떤 형태인지 조금 더 자세히 봅니다 - 이 자리만 잡으면 새 도구가 나와도 바로 어디에 속하는지 알 수 있습니다.

**① 편집기/IDE형 - 익숙한 코드 편집기 안에서**
변경 사항을 눈으로 보며 승인·취소할 수 있어 **심리적 문턱이 낮습니다.**
- 예: **Cursor**(전반적으로 무난), **Devin Desktop**(구 Windsurf), **GitHub Copilot**, **Cline**(오픈소스 확장)

**② 터미널/CLI형 - 검은 명령줄 화면에서 대화**
강력하고 확장이 깊지만 **명령줄에 익숙해야** 편합니다.
- 예: **Claude Code**(어려운 작업 추론에 강함), **OpenAI Codex CLI**, **Antigravity CLI**(agy), **Aider**·**opencode**(오픈소스·내 모델 골라 쓰기)

**③ 클라우드/비동기 에이전트형 - 과제를 던지면 알아서, 결과는 '초안'으로**
"이 이슈 고쳐 줘"를 맡기면 백그라운드에서 일하고 결과를 가져옵니다.
- 예: **Devin**, **GitHub Copilot 코딩 에이전트**, **OpenAI Codex(클라우드)**, **Google Jules**, **Google Antigravity**

### 역사 이야기: "세계 최초의 AI 소프트웨어 엔지니어" 논쟁 - Devin

2024년 3월, 한 스타트업(Cognition)이 **Devin** 을 공개하며 **"세계 최초의 AI 소프트웨어 엔지니어"** 라고 내세웠습니다. 프리랜서 플랫폼에서 실제로 일을 받아 돈을 번다는 시연 영상이 큰 화제가 됐죠.

그런데 곧 반론이 나왔습니다. 한 개발자가 그 영상을 뜯어보니 **홍보만큼 과제를 제대로 끝내지 못했고**, 잘 되는 장면만 골라 담았다는 지적이었습니다. 이후 독립 테스트에서도 성공률이 기대에 크게 못 미쳤습니다. "첫 AI 엔지니어"라는 **마케팅과 실제 성능 사이의 간극** 을 둘러싼 논쟁이 한동안 이어졌습니다.

> 교훈 두 가지 - ① 화려한 데모는 **가려서 볼 것.** "된다"는 시연과 "믿고 맡길 수 있다"는 전혀 다른 이야기입니다. ② 그럼에도 이 소동은 자율 코딩 에이전트 경쟁에 불을 붙인 **촉매** 였습니다. 과대광고와 진짜 발전은 늘 섞여 옵니다. (출처: The Register, 2025; IT Pro)

### 초보라면 어떤 형태부터, 그리고 무엇을 보고 고를까

**어떤 형태부터?**
- **만들기부터 배포까지 클릭으로** 끝내고 싶다면, 앞서 본 **브라우저형(Lovable·Bolt·Replit)** 이 문턱이 가장 낮습니다.
- **코드를 조금씩 보며 배우고 싶다면 편집기형** 이 좋습니다 - AI가 고친 걸 눈으로 보고 승인/취소하며 감을 익힐 수 있으니까요. 무료로 가볍게는 **GitHub Copilot 무료 티어**, 좀 더 본격적으로는 **Cursor**.
- **터미널형** 은 힘이 세지만 검은 화면이 익숙해진 뒤에 넘어가세요.
- **클라우드형** 은 "알아서 다 해 줌"이 매력이지만, 결과를 **검토·판단할 안목** 이 아직 부족한 초보의 첫 도구로는 통제감이 떨어집니다.

**고를 때 볼 기준 다섯 가지**
1. **모델을 고를 수 있나** - 한 회사 모델만 되나, 아무 모델이나 꽂나(오픈소스 계열이 자유로움).
2. **비용 구조** - 월 정액인가, "프로그램은 무료지만 쓴 만큼 API 요금"인가(후자는 많이 쓰면 확 오를 수 있음).
3. **오픈소스 여부** - 오픈소스면 그 도구가 서비스를 접어도 남이 이어받아 살릴 수 있습니다(실제로 2026년 한 오픈소스 편집기 확장의 종료가 이 리스크를 보여줬죠).
4. **터미널이 편한가** - 편하면 CLI형, 아니면 편집기형.
5. **맡길까 지켜볼까** - 한 줄씩 보며 승인하려면 편집기형, 통째로 맡기고 결과만 검토하려면 클라우드형.

> 참고 - 실무자들도 **하나만 쓰지 않습니다.** 어려운 건 터미널 에이전트, 일상 편집은 편집기, 가끔 백그라운드용으로 클라우드 에이전트를 **섞어 씁니다.** 여러분도 하나에 정착하기보다 **필요에 따라 갈아타는** 걸 목표로 두세요.

### 에이전트에게 컴퓨터를 맡길 때 - 안전과 권한

에이전트는 단순 답변기가 아니라 **실제로 파일을 지우고, 명령을 실행하고, 외부에 요청을 보낼 수 있는** 존재입니다. 그래서 좋은 도구는 위험한 행동 전에 **"이거 해도 될까요?"** 를 묻습니다. 문제는 확인 요청이 반복될수록 사람이 내용을 읽지 않고 "예"를 누르기 쉬워진다는 점입니다. 승인 창이 있다는 것만으로 안전해지지는 않습니다.

권한 방식은 도구마다 이름은 달라도 개념은 같습니다 - **매번 확인 / 편집만 자동 / 위험한 것만 막고 나머지 자동 / 아무것도 안 물어봄**. 마지막(전부 자동)은 **폐기해도 되는 실험 환경에서만** 쓰라고 공식적으로 경고합니다.

누구나 지킬 안전 습관 세 가지:

1. **중요한 폴더에서 '자동 수락'을 켜지 마세요.** 문서·사진이 든 폴더, 회사 자료 폴더에선 매번 확인 모드로.
2. **되돌릴 지점을 먼저 만드세요.** **Git이 추적하는 코드** 는 위험한 작업 전에 **저장(커밋)** 해 두면 그 상태로 되돌릴 수 있습니다. 하지만 추적하지 않는 파일이나 데이터베이스의 데이터까지 커밋으로 복구되는 것은 아닙니다. 이런 자료는 별도로 백업하세요.
3. 핵심 원칙 한 문장 - **"아무것도 안 물어봄"과 "중요한 것에 대한 전체 접근"을 동시에 켜지 마라.**

> **편함과 위험은 같이 옵니다.** 편해질수록 "이 작업이 뭘 건드리지?"를 한 번 더 보는 게 사람의 몫입니다.

### 개념 정리: 내 컴퓨터에서 도는 AI - 로컬 모델

지금 쓰는 클라우드 AI는 회사 서버에서 돌지만, **Ollama** 같은 도구를 쓰면 오픈 모델을 **내 컴퓨터에서 직접** 돌릴 수도 있습니다. 명령 한 줄이나 데스크톱 앱으로 모델을 내려받아 실행하죠. (출처: ollama.com)

장단점이 뚜렷합니다.

- **좋은 점(사생활·비용)** - 입력한 내용이 **기기 밖으로 안 나갑니다.** "클라우드 AI는 정책으로 사생활을 지키고, 로컬 AI는 구조로 지킨다"고들 합니다. 사용료가 없고 오프라인에서도 됩니다.
- **아쉬운 점(성능·사양)** - 집 컴퓨터에서 돌릴 수 있는 모델은 최신 클라우드 모델보다 대체로 뒤처지고, 어려운 추론엔 약합니다. 큰 모델은 고사양 GPU가 필요합니다.

> 결론 - **아직은 클라우드 모델이 무난합니다.** 다만 "계약서·개인정보처럼 민감한 자료를 다룰 땐 로컬, 어려운 작업엔 클라우드"라는 **섞어 쓰기** 감각을 알아 두면 좋습니다.

### 요즘 판: 열린 모델과 닫힌 모델

'모델 + 하네스'에서 모델 쪽에도 요즘 큰 판이 벌어지고 있습니다. 회사마다 태도가 갈려요. **Anthropic(Claude)** 은 모델의 가중치(모델의 '두뇌'에 해당하는 수치 뭉치. 공개하면 남이 그대로 받아 자기 컴퓨터에서 돌릴 수 있어요)를 **공개하지 않습니다.** "최고 성능을 유지하며 값을 받는다"는 전략이라, 반드시 이 회사 서버(API)로만 씁니다. 반대로 **OpenAI** 는 절충이에요. 최상위(GPT-5.x)는 닫아 두되, 한 급 낮은 **gpt-oss** 는 가중치를 풀어 "AI를 자기 컴퓨터에서 돌려야 하는 곳(병원·은행·정부)"을 붙잡습니다.

그런데 반전이 있습니다. 정작 **가장 센 '열린(오픈웨이트) 모델'은 중국에서** 나오고 있어요. DeepSeek, 알리바바의 Qwen, Moonshot의 Kimi, Z.ai의 GLM처럼 가중치를 공개하는 모델이 주목받고 있습니다. 다만 라이선스는 모델마다 다릅니다. **Kimi K3는 가중치 공개가 완료되었으며, MIT와 다른 Kimi K3 License를 따릅니다. 상업적 사용에는 조건이 있으므로 사용 전에 확인해야 합니다.** 이런 모델들은 코딩 실력이 프런티어(최첨단 최상위 모델)에 몇 점 차로 바짝 붙었고 일부 항목은 앞섭니다. 2026년 7월 나온 **Kimi K3**(2.8조 파라미터)는 독립 지능 지표에서 Claude Opus 4.8과 거의 동급(한 끗 위)이고, 프런트엔드 코드 아레나에서는 Claude Fable 5를 제치고 1위를 찍었습니다.

대략의 요금(2026년 7월, 100만 토큰당 · 빠르게 바뀜):

| 모델 | 입력 | 출력 | 가중치 공개 |
|---|---|---|---|
| Claude Opus 4.8 | $5 | $25 | 아니오 |
| Kimi K3 | $3 | $15 | 예(Kimi K3 License, 상업적 사용 조건 있음) |
| DeepSeek V4 | $0.5 미만 | 약 $1 안팎 | 예 |

다만 과장은 금물입니다. 가장 어려운 표준 벤치마크(SWE-bench Verified)의 최상단은 아직 프런티어 폐쇄 모델(Claude Fable 5 등)이 지키고 있고, 오픈 모델이 자기들이 낸 코딩 점수는 독립 검증이 진행 중입니다. '오픈웨이트 = 공짜'도 아니에요. Kimi K3를 직접 돌리려면 고성능 GPU 수십 장이 필요하고, 공개된 가중치를 쓰더라도 실행 비용과 라이선스 조건은 따로 확인해야 합니다. 게다가 앞서 봤듯 벤치마크 점수는 모델보다 하네스(스캐폴딩)가 더 좌우합니다.

요점은 하나예요. 이제 **모델은 갈아 끼우는 부품** 이고, 실력 격차가 좁혀지면서 **값·자체 실행(사생활)·특정 회사 종속** 이 진짜 선택 기준이 됩니다. 그래서 모델을 가리지 않는 도구(뒤에서 볼 omo 같은)가 뜨고, 서구 API 없이도 기관이 프런티어급을 자기 손으로 돌릴 길이 열리고 있죠. (출처: Artificial Analysis, artificialanalysis.ai; OpenRouter, "The open weight models that matter", 2026)

### 비용 이야기: 에이전트는 왜 '비쌀' 수 있나 - 토큰과 컨텍스트

에이전트가 강력한 데는 비용이 따릅니다. 앞서 본 루프(계획-실행-관찰-반성)를 여러 번 도는 동안 AI는 매번 상황을 다시 읽고 생각합니다. 그래서 **에이전트는 보통의 채팅보다 사용량(토큰)을 약 4배**, 여러 에이전트가 협업하면 약 15배까지 씁니다. '토큰'은 AI가 읽고 쓰는 글자 뭉치의 단위이자 곧 **요금의 단위** 입니다.

요금은 보통 **입력 토큰**(지시문·파일·이전 대화·도구 결과를 모델에 넣는 양)과 **출력 토큰**(모델이 새로 쓰는 양)을 따로 계산합니다. 그리고 한 번의 호출에서 모델이 참고할 수 있는 입력·출력의 총량에는 한도가 있는데, 이를 **컨텍스트 창(context window)** 이라고 합니다. 창이 크다고 공짜로 더 오래 기억하는 것은 아닙니다.

여기서 에이전트 비용이 왜 눈덩이처럼 커지는지가 보입니다. 에이전트는 한 작업도 여러 번 호출하고, i번째 호출은 앞선 대화와 도구 결과를 **다시 통째로** 포함합니다. 그러면 총 입력량은 대략 `Σ input_i`, 출력량은 `Σ output_i`가 됩니다. 매 단계에 누적 기록을 다시 넣으므로, 단계 수 n이 늘면 입력량은 선형보다 빠르게(단순 누적이라면 대략 n²에 가깝게) 커질 수 있습니다. 비용뿐 아니라 응답을 기다리는 시간도 그만큼 늘죠.

> 그래서 실전 규율은 하나입니다 - **한 번에 거대한 작업을 통째로 시키지 말고 작게 쪼개기.** 큰 파일·긴 로그는 필요한 부분만 주고, 작업이 끝나면 짧은 요약만 남깁니다. "토큰을 무조건 적게"가 아니라 **다음 판단에 필요한 증거만 남기는 것** 이 목표입니다('한 번에 하나만 바꾼다' 규율과도 통합니다). (출처: Anthropic, "How we built our multi-agent research system", 2025; "Effective context engineering")

### 작동 원리: 대화가 길어지면 왜 AI가 헤맬까 - 컨텍스트 관리

방금 본 **컨텍스트 창** 이 대화가 길어져 꽉 차기 시작하면 이상한 일이 벌어집니다 - 아까 정한 걸 잊고, 앞뒤가 안 맞고, 없는 파일명을 지어내기도 하죠. 이걸 반쯤 농담으로 **'컨텍스트 부패(context rot)'** 라 부릅니다. 창이 커져도, 입력이 길수록 성능이 떨어지는 경향은 여전합니다. (출처: Anthropic, "Effective context engineering")

그래서 실용 팁은 **"작업 단위로 대화를 정리하라"** 입니다. 서로 다른 작업의 지시가 한 대화에 쌓이지 않도록 구분하는 것이죠. 도구마다 방법은 있지만 개념은 둘입니다 - **요약해 압축하기**(한 작업이 끝난 시점) / **완전히 새 대화로 리셋하기**(전혀 다른 작업으로 넘어갈 때). 컨텍스트 파일(AGENTS.md)에 적은 규칙은 리셋해도 유지됩니다.

> 한 줄만 기억하면 - **한 가지 일이 끝나면 새 대화로 시작하는 게 오히려 결과가 좋습니다.** 하나의 긴 대화에 모든 걸 욱여넣지 마세요.

### 개념 정리: 큰 일은 여러 AI에게 나눠 - 서브에이전트

큰 일을 하나의 AI가 다 하는 대신, **여러 AI에게 나눠 맡기는** 방식이 있습니다. 지휘하는 AI가 하위 AI에게 "이것만 조사해와" 시키면, 하위 AI가 별도 공간에서 일한 뒤 **요약만** 가져오는 식이죠(이걸 **서브에이전트** 라 합니다). 더 나아가 여러 AI가 서로 대화하며 협업하는 **에이전트 팀** 도 있습니다.

강력하지만 대가가 있습니다 - 앞의 '토큰' 이야기 기억하시죠? **AI를 여러 개 돌리면 그만큼 토큰도 크게** 늘어납니다(여러 에이전트가 협업하면 보통 채팅의 십수 배까지). 그래서 실무에선 "지휘하는 AI는 똑똑한 모델, 잔일 하는 AI는 값싼 모델"로 **등급을 나눠** 비용을 아낍니다. (출처: Anthropic, "How we built our multi-agent research system")

> 감만 잡으면 - **"여러 AI에게 나눠 시킬 수 있지만, 편한 만큼 돈(토큰)을 훨씬 많이 쓴다."** 처음엔 하나의 에이전트로 충분합니다.

### 이 바닥은 정말 빠르게 바뀝니다

AI 코딩 도구의 이름·가격·무료 정책·회사는 짧은 기간에도 바뀝니다. 그래서 특정 도구의 인수·중단·대체 여부는 이 노트의 고정 사실로 외우기보다, **선택 직전에 공식 공지와 현재 저장소 상태를 확인할 정보** 로 다루는 편이 안전합니다. 비교 글은 출발점일 뿐, 최신성의 보증은 아닙니다.

그래서 **도구 이름을 외우는 건 의미가 적습니다.** 대신 오늘 배운 개념 - **에이전트 = 모델 + 하네스, MCP(연결), 스킬(반복 자동화)** - 을 잡아 두면, 어떤 새 도구가 나와도 "아, 이건 하네스가 이렇구나, MCP 되나?"로 빠르게 판단할 수 있습니다.

> **AI에게 고르게 하는 프롬프트** - "'브라우저에서 쓰고 싶고, 무료로 시작하고 싶다'는 조건에 맞는 AI 코딩 도구 3개를 장단점과 함께 추천하고, MCP·스킬 확장이 되는지도 알려줘. 각 정보의 확인 날짜와 공식 출처도 붙여줘."

## 스킬 너머 - 확장과 요즘 흐름

여기부터는 당장 필요하진 않습니다. 스킬·MCP에 익숙해지면 만나게 될, '한 걸음 더' 나가는 이야기예요. 지금은 이런 게 있다는 감만 잡아도 충분하고, 이 중 끌리는 것만 골라 읽어도 됩니다.

### 스킬·도구를 통째로 나누기 - Claude 플러그인

스킬 하나, MCP 연결 하나, 자주 쓰는 명령 하나. 이렇게 낱개로 쌓다 보면 남한테 "내 세팅 그대로 써 봐"라고 넘기기가 번거롭습니다. 그래서 Claude Code에는 이걸 하나로 묶는 '플러그인(plugin)'이 있습니다. 스킬, 서브에이전트(큰 일을 나눠 맡는 보조 AI), 자주 쓰는 슬래시 명령(/로 시작하는 단축 명령), 자동화 훅(정해진 순간에 자동 실행되는 동작), MCP 연결까지 한 꾸러미로 담아, 명령 한 줄이면 통째로 설치됩니다.

설치는 이렇게 합니다.

```bash
# 마켓플레이스(플러그인 모음)를 하나 등록하고
/plugin marketplace add anthropics/claude-plugins-community
# 거기서 원하는 플러그인을 설치
/plugin install <플러그인이름>@claude-community
```

Anthropic이 직접 고른 공식 모음(claude-plugins-official)과 사람들이 올린 커뮤니티 모음(claude-plugins-community)이 있고, 깃허브 주소나 내 폴더에서 바로 설치할 수도 있습니다. 필요할 때 켜고 안 쓸 때 끄면 되는데, 많이 켤수록 AI가 읽어야 할 맥락이 늘어 느려지니 '지금 쓸 것만' 켜는 게 좋습니다.

플러그인이 진짜 빛나는 건 팀에서입니다. 보통 좋은 세팅은 잘하는 한 사람 머릿속(이나 노트북)에만 있죠. 그걸 플러그인으로 묶어 두면 "우리 팀 코드 리뷰는 이렇게", "배포는 이 순서로" 같은 노하우가 설치 한 번으로 모두의 것이 됩니다. 버전이 붙어 있어 만든 사람이 고치면 다 같이 업데이트되고, 문제가 생기면 이전 버전으로 되돌리기도 됩니다. (출처: Anthropic, "Claude Code plugins" 공식 문서, 2026)

### 한 걸음 더 - 에이전트 팀을 통째로 얹기

지금까지는 스킬 하나, MCP 하나를 손으로 붙였죠. 여기서 더 나아가면, 여러 전문 에이전트와 스킬·명령을 **한꺼번에** 얹어 Claude Code를 '지휘자'로 만드는 오픈소스 레이어도 있습니다. 큰 일이 들어오면 지휘자가 알아서 조사·구현·검증 담당에게 나눠 맡기고 병렬로 굴리는 식이죠. 여러 AI에게 일을 나눠 맡기는 서브에이전트를 아예 '팀' 규모로 키운 셈입니다.

대표적으로 두 갈래가 있습니다. **oh-my-claudecode**(<https://github.com/Yeachan-Heo/oh-my-claudecode>)는 Claude Code에 플러그인으로 얹는 '팀 우선' 오케스트레이션 레이어입니다. 마켓플레이스에 저장소를 등록하고 설치하면 수십 개의 전문 에이전트와 스킬이 한 번에 딸려 옵니다.

```bash
/plugin marketplace add https://github.com/Yeachan-Heo/oh-my-claudecode
/plugin install oh-my-claudecode
```

**oh-my-openagent**(<https://github.com/code-yeongyu/oh-my-openagent>, 줄여서 omo)는 한발 더 나아가 모델을 가리지 않습니다. Claude든 GPT든 Gemini든 한 인터페이스 뒤에 묶어, "이 잔일엔 값싼 모델, 이 판단엔 제일 똑똑한 모델"처럼 골라 쓰게 하죠. 특정 회사 도구에 묶이고 싶지 않을 때의 선택입니다.

> 첫날부터 손댈 도구는 아닙니다. 그래도 알아 둘 값어치가 있어요. "AI 하나에게 다 시키기"에서 "여러 전문 AI를 지휘하기"로 넘어가는 흐름의 끝에 이런 레이어가 있고, 이게 오늘 배운 스킬·플러그인·서브에이전트를 차곡차곡 쌓아 올린 결과라는 것. 익숙해지면 자연스레 이쪽으로 올라오게 됩니다. (출처: oh-my-claudecode·oh-my-openagent 공식 저장소, 2026)

### 한 걸음 더 - '먼저 적고 시키는' 명세 주도 개발

지금까지 본 스킬과 AGENTS.md에는 공통점이 하나 있습니다. 시키기 전에 '무엇을, 어떻게'를 먼저 글로 적어 둔다는 것. 이 발상을 프로젝트 전체로 키운 게 요즘 부쩍 뜨는 명세 주도 개발(spec-driven development)입니다.

배경은 이렇습니다. 아무 설명 없이 "만들어 줘"만 던지는 이른바 '바이브 코딩'은 빠르지만, AI가 슬금슬금 내 의도에서 벗어나거나 없는 기능을 지어내기 십상입니다. 그래서 코드를 짜기 전에 요구사항·설계·할 일 목록을 먼저 문서로 정하고, AI가 그 문서를 보며 단계대로 만들게 하자는 거죠. 한마디로 "코드가 정답이 아니라, 내 의도(명세)가 정답이다"로 무게중심을 옮기는 겁니다. 우리가 앞서 과제로 써 본 PRD가 바로 그 '명세'의 출발점이었고요.

도구도 이미 많습니다. 깃허브가 내놓은 오픈소스 스펙킷, [Spec Kit](https://github.com/github/spec-kit)은 '원칙 → 명세 → 계획 → 할 일 → 구현'을 정해진 순서(각 단계가 `/speckit.specify` 같은 명령)로 밟게 해 줍니다. 깃허브 별이 12만 개를 넘을 만큼 인기이고, 여러 AI 코딩 도구에서 두루 쓰이죠. 아마존은 아예 명세부터 쓰게 만드는 편집기 Kiro를 내놨고요. 더 가볍게 시작하고 싶다면 [OpenSpec](https://github.com/Fission-AI/openspec)도 많이 쓰입니다. 명령 한 줄로 설치되고, 특히 이미 굴러가는 프로젝트를 조금씩 고쳐 나갈 때 편하다는 평이죠. 방식은 조금씩 달라도 뿌리는 같아요. 짧게라도 먼저 적고 시키면 AI가 덜 헤매고, 결과도 고쳐 쓰기 좋습니다.

다만 직접 써 본 사람들 얘기를 들어 보면 이게 만능은 아닙니다. 그래서 요즘 굳어지는 결론은 '섞어 쓰기'예요. 가볍게 탐색하거나 시제품을 만들 땐 그냥 대화로(바이브 코딩), 오래 유지할 중요한 걸 제대로 지을 땐 명세로. 한 줄로 남기면, 명세 주도 개발의 진짜 값어치는 도구 자체가 아니라 '만들기 전에 한 번 더 생각하게 만든다'는 데 있습니다. (출처: GitHub Spec Kit·AWS Kiro 공식 문서; 실무자 후기 종합, 2026)

### 쌓일수록 강해지기 - 컴파운드 엔지니어링

앞의 AGENTS.md·스킬에는 숨은 힘이 하나 있습니다. 한 번 적어 두면 다음 작업이 더 수월해진다는 것. 이 아이디어를 아예 방법론으로 밀어붙인 게 컴파운드 엔지니어링(compound engineering)입니다. 핵심은 '일할 때마다 그 경험을 규칙·스킬로 남겨, 다음 일이 이전 일 위에 쌓이게' 하는 거예요. 버그를 한 번 고치면 그 교훈을 AGENTS.md에 적어 두고, AI가 다음부터 자동으로 지키게 하는 식이죠. 개발사 Every가 이 방식으로 소수 인원이 여러 제품을 굴린다고 알려졌고, 같은 이름의 Claude Code 플러그인도 인기입니다(코드 리뷰를 여러 전문 에이전트가 병렬로 나눠 보는 기능이 대표적이죠).

이걸 도구로 구현한 예가 Nous Research의 Hermes입니다. 보통 AI는 대화창을 닫으면 배운 걸 잊지만, Hermes는 기억을 남기고 스스로 스킬을 만들어 오래 쓸수록 똑똑해지도록 설계됐습니다(모델은 갈아 끼울 수 있고요). 만든 쪽은 '진짜 열쇠는 더 똑똑한 모델이 아니라 모델을 감싸는 더 똑똑한 껍데기(하네스)'라고 말하는데, 앞서 배운 '모델 + 하네스'의 하네스를 끝까지 밀어붙인 셈입니다. (출처: Every, "Compound Engineering" https://every.to/guides/compound-engineering; Nous Research, Hermes Agent https://github.com/NousResearch/hermes-agent)

### 프롬프트를 넘어 루프로 - 루프 엔지니어링과 'Ralph'

앞에서 에이전트가 '계획-실행-관찰-반복' 루프를 돈다고 했죠. 요즘은 그 루프를 사람이 직접 설계하는 흐름이 뜹니다. 대표가 개발자 제프리 헌틀리가 퍼뜨린 'Ralph'예요(이름은 심슨 캐릭터 랠프 위검에서 땄습니다. '단순해 보이는데 진짜 되더라'는 뜻으로 붙였고, 유래와 실전 방법은 헌틀리의 원문 https://ghuntley.com/ralph 에 잘 정리돼 있습니다). 방법은 놀랄 만큼 단순합니다. 같은 지시(스펙 = 앞서 본 명세 파일)를 계속 다시 주고, 한 번에 한 가지만 하고 커밋하기를 목표를 이룰 때까지 반복합니다. 진행 상황은 대화 기억이 아니라 파일과 git에 쌓아 두고요. (헌틀리의 원래 방식은 명령줄에서 에이전트를 매번 새로 호출하는 bash 루프라, 한 바퀴가 새 프로세스로 시작됩니다. 다만 '깨끗한 대화 리셋'이 Ralph의 정해진 기능은 아니고, 같은 세션에 지시만 다시 넣는 구현도 많습니다.)

왜 하나의 긴 대화 대신 이렇게 할까요? 앞서 본 '컨텍스트 부패'를 떠올리면 됩니다. 대화가 길어질수록 AI가 헤매니, 상태를 대화 기억에 쌓지 않고 스펙·AGENTS.md 같은 고정 파일과 git에 두는 겁니다. 그러면 어느 바퀴에서 시작하든 AI가 같은 안내 파일을 근거로 삼죠. 2023년 AutoGPT가 '무한 반복'으로 실패했던 자리를 '멈출 기준(스펙) + 파일에 쌓는 진행'으로 다시 푼 셈입니다. 헌틀리는 이 방식으로 프로그래밍 언어 하나를 약 300달러어치 호출로 만들어 화제가 됐습니다.

이 흐름이 굳어져 이제 도구에 아예 들어왔습니다. Claude Code에는 목표만 주면 알아서 도는 `/goal` 명령이 생겼고, Anthropic과 OpenAI가 '루프 만드는 법'을 공식 문서로 냈죠. Claude Code를 만든 보리스 체르니는 '나는 이제 프롬프트하지 않는다, 루프를 돌리고 그 루프가 대신 프롬프트한다'고까지 말합니다. 다만 안전장치가 없으면 무한 반복·엉뚱한 목표·요금 폭발이 흔하니, 멈출 기준과 검토는 여전히 사람 몫입니다. (출처: Geoffrey Huntley, "ralph"; Anthropic, "Getting started with loops", 2026)

### 요즘 흐름 - MCP에서 '스킬 + CLI'로

MCP로 도구를 연결하면 도구의 이름·설명·입력 형식을 AI에게 알려 줘야 합니다. 이 설명을 처음부터 모두 읽는 방식에서는 컨텍스트(기억 공간)를 많이 쓸 수 있어요. 서버와 도구가 많으면 설명만 수만 토큰에 이르는 사례도 있지만, 실제 부담은 도구 수와 설명 길이, 필요한 도구만 불러오는 기능의 사용 여부에 따라 다릅니다. 2026년에는 이미 있는 명령줄 도구(CLI)를 스킬과 함께 활용하자는 흐름도 주목받고 있습니다.

예를 들어 깃허브 작업에는 `gh`라는 공식 CLI가 있습니다. 설치와 로그인을 마쳤다면 AI에게 "`gh`로 이슈 목록을 가져와"라고 요청할 수 있죠. 여기에 **짧은 스킬 파일 하나** 로 언제 어떤 명령을 쓸지 알려 줄 수 있습니다. CLI 약 1천 토큰 대 MCP 약 4만 토큰, 800토큰짜리 스킬 대 2만 8천 토큰짜리 MCP 설명서라는 비교 수치가 소개되기도 합니다. 다만 여기서는 원문과 측정 조건을 확인할 수 없으므로 일반적인 성능 차이로 받아들이지 마세요. CLI도 명령의 출력과 오류를 읽는 비용이 있고, 스킬을 붙인다고 항상 더 빠르고 정확해지는 것은 아닙니다.

도구 설명을 불러오는 방식도 바뀌고 있습니다. Claude Code에는 필요한 MCP 도구를 **검색해서 불러오는** 툴 서치 기능이 있습니다. 적용 여부는 버전과 설정에 따라 확인해야 하지만, 모든 설명을 한꺼번에 읽는 부담을 줄이는 방법입니다. 따라서 MCP와 CLI를 비교할 때는 같은 작업에서 어떤 설명과 출력이 전달되는지 함께 살펴보세요.

그렇다고 MCP가 한물간 건 아닙니다. 여러 사람이 같이 쓰고 접근 권한·기록이 중요한 회사 환경에서는 MCP의 표준화된 관리가 여전히 강하거든요. 그래서 요즘 정리는 "셋 중 하나가 아니라, 스킬·MCP·CLI를 함께"입니다. 이렇게 기억하면 됩니다. 이미 있는 CLI로 해결할 수 있는지 먼저 살피고, 여러 도구의 연결 방식을 통일해야 한다면 MCP도 검토하세요. 어느 쪽이든 인증과 권한 설정은 따로 확인해야 합니다. (출처: Anthropic, "Code execution with MCP", 2025; 2026년 비교 수치는 원문·측정 조건 미확인)

## 실습 - Todo 앱을 'AI가 아는' 앱으로

할 일을 저장하는 앱을 재료로, 범용 AI를 **'내 앱을 아는 전용 조수'** 로 바꿔 봅시다. 앱이 없어도 아래 예제 앱부터 만들면 실습할 수 있습니다.

방향은 이 강의가 앞서 짚은 흐름 그대로입니다 - 무거운 MCP부터 세우기보다, **앱에 명령줄 도구(CLI)를 갖추고 그 사용법을 규칙 파일에 적어 두면**, AI가 그 명령을 직접 돌려 내 데이터를 읽고 다룹니다. 거기에 스킬로 반복 작업을 외워 두면 끝이죠.

**이 실습에서 하는 것**

- 앱에 **CLI**(`todo list`·`summary` 같은 명령)를 갖추고, `AGENTS.md`(규칙 파일)에 사용법을 적는다.
- AI에게 그 **CLI를 실제로 돌리게** 해 내 할 일 데이터를 읽는다(이때 **권한 승인** 을 관찰하고, 코드 수정 전 **커밋** 으로 Git이 추적하는 코드의 되돌릴 지점을 만든다).
- `SKILL.md`를 하나 만들어, 매번 반복하던 정리 작업을 **한 번 적어 재사용** 한다.
- (덤) 같은 걸 **MCP로 붙이면** 어떻게 다른지 짧게 체험한다.

준비물은 앱 프로젝트를 열고 명령을 실행할 수 있는 에이전트형 도구(Claude Code·Cursor 등)입니다. 앱은 아래 두 갈래 중 하나를 골라 준비하세요.

**(a) 작은 로컬 예제 앱으로 시작하기**

실제 서비스 데이터와 분리된 빈 폴더를 열고 AI에게 이렇게 요청하세요.

> Node.js로 로컬 파일 todo.db에 할 일을 저장하는 아주 작은 예제 앱을 만들어 줘. 이 폴더 안에서만 작업하고, 실제 계정이나 클라우드 DB는 연결하지 마. 할 일 추가·목록 조회·완료 처리가 가능한 간단한 웹 화면을 만들고, 각 할 일에 고유 번호·내용·완료 여부·완료 시각을 저장해 줘. 설치와 실행 명령을 알려 주고, 가상 할 일을 넣어 앱을 껐다 켜도 남는지 확인해 줘. 명령줄용 진입점은 아직 만들지 말고, 저장·조회 로직을 별도로 재사용할 수 있게 구성해 줘.

앱을 실행해 화면에서 가상 할 일을 추가하고 완료해 보세요. 앱을 다시 열었을 때 데이터가 남으면 준비된 것입니다.

**(b) 기존 클라우드 DB 앱으로 시작하기**

Supabase 등 클라우드 DB에 할 일을 저장하는 앱이 있다면 그대로 사용할 수 있습니다. 단, 프로젝트 코드를 열 수 있고, 터미널에서도 기존 인증 방식과 권한으로 DB에 접근할 수 있어야 합니다. 실제 사용자 데이터가 아닌 테스트용 프로젝트나 분리된 테스트 데이터로 진행하세요. 인증 정보는 환경 변수로 전달하고, 규칙 파일·스킬·Git 저장소에는 넣지 않습니다. 접속을 위해 데이터 접근 제한을 풀거나 관리자 키를 무작정 사용하지 마세요. 안전한 테스트 환경이나 접속 방법을 마련하기 어렵다면 (a)를 선택하세요.

두 갈래 모두 같은 실습을 합니다. 차이는 저장 위치뿐입니다. 로컬 앱은 `todo.db`, 클라우드 앱은 기존 DB의 테스트 데이터를 CLI로 읽고, 그 결과를 스킬에서 재사용합니다.

### 1단계 - 앱에 '명령줄 사용법(CLI)'을 붙이기

준비한 Todo 앱을 화면(웹)으로 쓰고 있다면, AI를 위한 또 다른 입구를 붙여 봅시다. 그런데 AI가 다루기 가장 편한 창구는 화면이 아니라 **명령줄(CLI)** 이에요. 화면은 사람이 클릭하라고 있는 거라 AI가 붙잡기 번거롭지만, 명령줄은 "명령어를 쳐서 결과를 글로 받는" 방식이라 AI가 그대로 실행하고 읽기 좋거든요.

다행히 앱을 새로 만들 필요는 없습니다. 할 일을 추가·조회·완료하는 **로직은 이미 앱 안에 들어 있으니**, 그걸 터미널에서 부를 수 있는 **작은 입구(진입점) 하나만** 더 얹으면 됩니다. 데이터 저장 위치도 유지합니다. 로컬 예제 앱은 `todo.db`를, 클라우드 DB 앱은 준비한 테스트 데이터를 그대로 사용합니다. 같은 데이터를 여는 문을 하나 더 다는 셈이죠.

CLI를 한 줄로 정리하면, 프로그램에 "무엇을 할지"를 화면 클릭 대신 **명령어 + 값** 으로 전달하는 방식입니다. 예를 들어 `node todo.js add "우유 사기"`에서 `add`가 명령, `"우유 사기"`가 넘겨주는 값이에요.

그래서 AI에게 이렇게 부탁합니다 - *"우리 Todo 앱의 **기존 로직을 재사용해서** 터미널에서 쓸 `todo` 명령을 만들어 줘. `add`(추가)·`list`(목록)·`done`(완료)·`summary`(오늘 완료 요약) 네 개면 돼. 데이터는 현재 앱의 저장소를 그대로 사용해 줘. 로컬이면 todo.db를, 클라우드 DB면 기존 인증·접근 권한을 유지한 테스트 데이터를 사용하고, 앱 코드를 갈아엎지 말고 얇은 진입점만 더해 줘."* 그러면 이렇게 동작합니다(이하 편의상 `todo …`로 줄여 씁니다. AI가 실제로 만든 파일명은 `cli.js`처럼 다를 수 있으니, 나온 이름으로 바꿔 쓰면 됩니다).

```bash
$ node todo.js list
[x] 1. 발표 자료 초안 작성하기
[ ] 4. 헬스장 PT 예약 변경 문의하기
```

(앱이 파이썬이면 `python todo.py list`처럼, 쓰는 언어에 맞는 진입점이 생깁니다. 형태만 다를 뿐 원리는 같아요.) 한 가지 - `summary`가 '오늘 완료'를 세려면 완료한 **시각** 이 데이터에 있어야 합니다. 현재 앱에 완료 날짜가 없으면, AI에게 "완료 시각도 기록하게 해 줘"를 함께 부탁하세요.

이제 이 사용법을 앱 폴더 **맨 위(루트)** 의 규칙 파일에 5～10줄로 적어 둡니다. 파일 이름은 도구를 탑니다 - **Cursor** 등은 `AGENTS.md`를 그대로 읽고, **Claude Code** 는 버전과 설정에 따라 `CLAUDE.md`와 `AGENTS.md`를 자동으로 읽는 조건이 다릅니다. 아래 내용을 `CLAUDE.md`에 적거나, `CLAUDE.md`에서 `@AGENTS.md`로 불러오는 방식으로 준비하고 실제로 읽는지 확인하세요.

아래는 로컬 예제 앱의 규칙 파일입니다. 클라우드 DB 앱이라면 저장 위치와 보호 대상을 실제 테스트 DB에 맞게 바꾸고, 인증 정보는 환경 변수에서 읽는다고 적으세요. 파일명과 명령도 실제 생성된 CLI에 맞춥니다.

```markdown
# Todo 앱

## 개요
할 일을 추가·완료·조회하는 1인용 앱.

## 명령
- 목록: node todo.js list
- 추가: node todo.js add "<내용>"
- 완료: node todo.js done <번호>
- 오늘 완료 요약: node todo.js summary

## 규칙
- 데이터는 todo.db에 저장한다. UI 문구는 존댓말로.

## 하지 말 것
- todo.db(실데이터)를 직접 지우거나 덮어쓰지 말 것.
```

다 적었으면 **새 대화를 열고** "이 프로젝트 어떻게 실행해?"라고 물어보세요. AI가 방금 적은 명령을 **먼저 읽고 답하면** 성공입니다. 대화를 리셋해도 이 파일은 자동으로 읽히니, 같은 설명을 두 번 안 해도 됩니다.

### 2단계 - AI에게 그 CLI를 실제로 돌리게 하기

이제 진짜 재미있는 부분입니다. AI에게 이렇게 시켜 보세요 - **"오늘 할 일 목록 좀 보여 줘."** 규칙 파일을 읽은 AI는 스스로 `node todo.js list`를 **실행** 해서 그 출력을 가져옵니다. 복사·붙여넣기 없이, AI가 내 앱을 직접 돌려 데이터를 읽는 거죠. 소스 코드가 아니라 **실제 할 일 데이터** 를 보는 것이 핵심입니다.

**여기서 권한을 관찰하세요.** 도구의 권한 설정에 따라 실행 전에 승인창이 뜨거나, 이미 허용한 명령이 바로 실행될 수 있습니다. 먼저 읽기 전용인 `list`·`summary`의 실행 대상과 내용을 확인하세요.

거부 연습은 실데이터가 아니라 **버려도 되는 더미 파일** 로 합니다. 앱 폴더에 `연습용-삭제해도-됨.txt`를 만들고 아무 의미 없는 문장을 적으세요. 자동 승인을 끄고 파일 삭제 전에 확인을 받는 설정인지 살핀 뒤, AI에게 "연습용-삭제해도-됨.txt만 지워 줘"라고 요청합니다. 승인창에서 대상 파일을 확인하고 **거부** 하세요. 창이 뜨지 않고 실행됐다면 권한 설정을 다시 확인합니다. 이때도 실제 DB 파일이나 클라우드 데이터를 삭제 대상으로 바꾸지 마세요.

코드를 수정하기 전에는 변경 사항을 확인하고 `git commit`으로 되돌릴 지점을 만들어 둡니다. 이 방법이 보호하는 것은 **Git이 추적하는 코드** 입니다. `todo.db` 같은 데이터 파일이나 클라우드 DB의 내용까지 복구해 주지는 않으므로, 데이터는 별도의 백업과 테스트 환경으로 보호해야 합니다.

> 왜 CLI로 할까요? 앞서 본 '요즘 흐름' 그대로입니다 - AI는 `node`·`git` 같은 명령을 이미 알고 있어서, 무거운 연결을 새로 세우지 않고 **있는 명령을 그냥 부르는** 게 가볍고 빠릅니다.

### 3단계 - 반복 작업을 스킬로 외워 두기

> 이 단계는 Claude Code 기준입니다. Cursor 등 다른 도구는 스킬 위치가 다르니 개념만 봐도 됩니다.

매번 똑같이 시키는 일이 있죠 - "오늘 완료한 것만 뽑아 커밋 메시지 형식으로 정리해 줘" 같은. 이걸 스킬로 만들어 둡니다. 앱 폴더의 `.claude/skills/todo-summary/SKILL.md`에 이렇게 적으면 됩니다(Claude Code는 이 경로의 스킬을 자동으로 찾습니다).

```markdown
---
name: todo-summary
description: 오늘 완료한 할 일을 커밋 메시지 형식으로 요약할 때 사용
---

# 오늘 완료 요약

`node todo.js summary`를 실행하고, 그 출력을 그대로 보여준다.
```

핵심은 스킬이 직접 데이터를 뒤지는 게 아니라, **앞서 만든 CLI를 대신 불러 준다** 는 것입니다. 그다음 **다른 대화에서** "요약 스킬 써서 오늘 한 일 정리해 줘"라고만 하면, AI가 이 스킬을 알아서 꺼내 `todo summary`를 돌리고 결과를 정리합니다 - 앞서 배운 **점진 공개** 가 실제로 동작하는 순간이죠. (스킬 폴더를 처음 만들었다면 도구를 한 번 재시작해야 인식됩니다.)

### 덤(시간 남으면) - 같은 걸 MCP로 붙이면?

CLI 대신 MCP로도 붙일 수 있습니다. 파일시스템 MCP 서버를 하나 등록하면 AI가 앱 폴더를 읽죠 - 설정은 대개 이렇게 생겼습니다.

```json
{
  "mcpServers": {
    "todo-files": {
      "command": "npx",
      "args": ["-y", "@modelcontextprotocol/server-filesystem", "/내-Todo앱-경로"]
    }
  }
}
```

붙여 두면 "우리 앱 소스에서 할 일을 어떻게 저장하는지 설명해 줘" 같은 걸 시켜 볼 수 있습니다. 다만 직접 해 보면 차이가 느껴질 거예요. 이 서버는 **파일(소스)** 을 읽는 데는 좋지만, 데이터가 담긴 `todo.db`는 이진 파일이라 **내용까지 세어 주진 못합니다.** 방금 CLI로 한 "오늘 완료 몇 건" 같은 건 결국 앱의 명령(`todo summary`)을 돌려야 나오죠.

> 그래서 오늘 배운 판단이 여기서 그대로 쓰입니다 - **혼자 작은 앱이면 CLI + 스킬이 가볍고, 여럿이 쓰고 권한·기록을 챙겨야 하면 MCP.** 도구가 아니라 상황이 고릅니다.

### 정리 - 오늘의 결과물

아래를 갖추면 이 실습은 끝입니다.

- [ ] 앱에 CLI(`list`·`summary` 등) + 규칙 파일(`AGENTS.md`, Claude Code면 `CLAUDE.md`)에 사용법
- [ ] AI가 그 CLI를 **직접 실행** 해 내 할 일 데이터를 읽음(더미 파일 삭제 요청의 승인창을 **거부**, 코드 수정 전 `git commit`; 데이터는 별도 보호)

### 과제 - 내가 실제로 쓸 스킬 하나, GitHub에 올리기

오늘 배운 개념이 내 것이 되려면 직접 하나 만들어 봐야 합니다. **내가 매번 AI에게 반복해서 시키던 일** 을 스킬 하나로 만들어, 남이 봐도 쓸 수 있게 GitHub에 올리세요. 제출은 **저장소 링크 하나** 입니다.

**1. 스킬 후보 고르기**

"AI한테 또 이걸 설명하고 있네" 싶은 순간이 후보입니다. 예:
- 코드를 고칠 때마다 우리 팀 린트·컨벤션을 다시 알려 주는 일
- 작업을 끝내고 항상 돌리는 정리 (커밋 메시지·변경 요약)
- 회의록·안내문을 우리 팀 양식으로 바꾸는 일

**2. 저장소를 만들고 스킬 쓰기**

먼저 스킬을 실제로 쓸 프로젝트를 고릅니다. 이곳이 **적용 대상 프로젝트** 입니다. 별도로 스킬을 공유·제출할 GitHub 저장소를 하나 만듭니다(예: `my-skills`). 이곳은 **제출용 저장소** 이며, 여기에 파일을 올리는 것만으로 다른 프로젝트에 적용되지는 않습니다. 그 안에 스킬 폴더와 `SKILL.md`를 만듭니다. Claude Code 기준 경로는 `.claude/skills/<스킬이름>/SKILL.md`이고, 다른 도구는 각 도구의 스킬 위치에 둡니다. 파일은 프론트매터(`name`·`description`)와 본문으로 씁니다.

```markdown
---
name: commit-summary
description: 오늘 작업을 우리 팀 커밋 메시지 양식으로 요약할 때 사용
---

# 커밋 메시지 요약

1. `git diff`로 바뀐 부분을 확인한다.
2. 바뀐 것을 주제별로 묶어 한 줄씩 요약한다.
3. 양식은 `[범위] 한 일 (#이슈)` 로 맞춘다.
```

스킬을 작성한 뒤에는 제출용 저장소의 스킬 폴더를 **적용 대상 프로젝트** 의 `.claude/skills/<스킬이름>/`에 복사하세요. 다른 도구라면 그 도구가 안내하는 프로젝트용 위치에 설치합니다. 그 프로젝트를 연 상태에서 새 대화를 시작해 스킬을 호출하고 결과를 확인하세요. 위 예시의 `git diff`는 `my-skills`가 아니라 **변경을 요약할 프로젝트의 루트** 에서 실행해야 합니다. `git diff`는 아직 스테이징하지 않은 변경을 보여 주므로, 해당 변경이 있는 상태에서 시험하세요. 로컬 예제 앱 프로젝트에서 시험해도 됩니다.

**3. README로 '왜' 쓰기**

저장소 최상단 `README.md`에 **왜 이 스킬을 만들었는지** 를 한 장으로 씁니다. "무슨 불편이 있었고, 스킬이 어떻게 없앴는지"가 한눈에 들어와야 합니다. 적용 대상 프로젝트에 복사·설치할 위치, 호출 문장, 실제 확인한 결과도 함께 적으세요. 인증 정보나 개인 데이터는 포함하지 않습니다.

**4. 올리고 링크 제출**

push하고 저장소 링크를 제출합니다. 보너스로 친구에게 링크를 주고 "이 스킬 써 봐" 했을 때, 따로 설명하지 않아도 돌아가면 성공입니다.
